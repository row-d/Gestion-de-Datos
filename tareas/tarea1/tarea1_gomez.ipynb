## Limpieza de datos con justificación
  - Detectar valores NaN en cada DataFrame (usar .isna().sum()).
  - Imputar el promedio faltante con la media de la carrera correspondiente. Justificar en markdown por qué esta estrategia es razonable.
  - Imputar asistencia_pct faltante con la mediana de la sede.
  - Verificar que después de la imputación no quedan NaN.


```python
alumnos_df.isna().sum()[lambda id: id > 0]
inscripciones_df.isna().sum()[lambda id: id > 0]
sedes_df.isna().sum()[lambda id: id > 0]

# promedio
alumnos_sin_asistencia_i = alumnos_df.loc[alumnos_df.promedio.isna()].index

alumnos_df.loc[alumnos_sin_asistencia_i]
alumnos_df["promedio"] = alumnos_df.promedio.fillna(alumnos_df.promedio.mean())
# alternativa
# alumnos_df.loc[alumnos_sin_promedio_i, "promedio"] = alumnos_df.promedio.mean()
alumnos_df.loc[alumnos_sin_asistencia_i]


# asistencia_pct

alumnos_sin_asistencia_i = alumnos_df.loc[alumnos_df.asistencia_pct.isna()].index
mediana_asistencia_sedes = alumnos_df.groupby("sede")["asistencia_pct"].transform("median")

alumnos_df.loc[alumnos_sin_asistencia_i]
alumnos_df["asistencia_pct"] = alumnos_df["asistencia_pct"].fillna(mediana_asistencia_sedes)
alumnos_df.loc[alumnos_sin_asistencia_i]
```

**Posible justificación para "imputar" con la media de la carrera**:
    Hubo errores en el sistema o el profesor olvido/perdio la asistencia, y no hay datos suficientes para recuperar el porcentaje de asistencia del alumno. Por lo tanto la universidad debera ser responsable de dar una solucion aceptable para que el alumno cumpla con el porcentaje de asistencia para aprobar.
    Tambien se puede ver que si el promedio > 0, significa que al menos asistió a las evaluaciones, por lo tanto asistencia_pct no podria ser NaN.

## Ingeniería de variables
  - Crear una columna categoria_rendimiento en alumnos: 'destacado' si promedio ≥ 6.0, 'aceptable' si 5.0 ≤ promedio < 6.0, 'bajo' si < 5.0.
  - Crear una columna promedio_final en inscripciones = (nota1 + nota2 + nota3) / 3.
  - Verificar el resultado con head().


```python
def categoria_rendimiento_handler(promedio: float):
    if promedio >= 6.0:
        return "destacado"
    if promedio >= 5.0 and promedio < 6.0:
        return "aceptable"
    return "bajo"


inscripciones_df["promedio_final"] = (
    inscripciones_df.nota1.fillna(0)
    + inscripciones_df.nota2.fillna(0)
    + inscripciones_df.nota3.fillna(0)
) / 3

alumnos_df["categoria_rendimiento"] = alumnos_df["promedio"].transform(
    categoria_rendimiento_handler
)
alumnos_df.head()
inscripciones_df.loc[inscripciones_df["promedio_final"].isna()]
```

## Uniones, groupby y pivot_table
  - Unir alumnos con sedes mediante merge para agregar la región a cada alumno.
  - Unir el resultado con inscripciones para tener todo en una sola tabla.
  - Agrupar por sede y carrera y calcular el promedio_final medio (groupby).
  - Construir un pivot_table con filas = sede, columnas = carrera, valores = promedio_final medio.


```python
alumnos_region_df = alumnos_df.merge(sedes_df, on="sede")
alumnos_region_inscripciones = alumnos_region_df.merge(
    inscripciones_df, left_on="id", right_on="id_alumno"
)
alumnos_region_inscripciones.head()

alumnos_region_inscripciones.groupby("sede").promedio_final.mean()
alumnos_region_inscripciones.groupby("carrera").promedio_final.mean()

alumnos_region_inscripciones.pivot_table(
    values="promedio_final", index="sede", columns="carrera", aggfunc="mean"
)
```

## Fechas, filtros y exportación
  - Convertir fecha_inscripcion a datetime.
  - Filtrar los alumnos cuya fecha_inscripcion sea posterior al 01/05/2025.
  - Exportar ese subconjunto (con columnas id, nombre, carrera, sede, promedio, fecha_inscripcion) a un archivo llamado reporte_filtrado.csv.


```python
alumnos_df["fecha_inscripcion"] = alumnos_df.fecha_inscripcion.transform(
    lambda fecha: platano.to_datetime(fecha, format="%Y-%m-%d")
)

alumnos_fecha = alumnos_df.loc[
    alumnos_df["fecha_inscripcion"] > platano.to_datetime("2025-05-01", format="%Y-%m-%d")
]

alumnos_fecha[
    ["id", "nombre", "carrera", "sede", "promedio", "fecha_inscripcion"]
].to_csv(output_data_dir / "reporte_filtrado.csv")
```
